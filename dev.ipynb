```python
from pdf2image import convert_from_path
path=r"/home/nikil/Documents/Projects/Wingman_Edge/Resume_AI_Engineer.pdf"
images = convert_from_path(path)

# save pages
paths = []
for i, img in enumerate(images):
    path = f"page_{i}.png"
    img.save(path, "PNG")
    paths.append(path)
```

```python
import ollama

results = []

for path in paths:
    response = ollama.chat(
        model="glm-ocr:q8_0",
        messages=[
            {
                "role": "user",
                "content": "Extract all text from this document",
                "images": [path]
            }
        ]
    )
    
    results.append(response["message"]["content"])

full_text = "\n\n".join(results)
print(full_text)
```
